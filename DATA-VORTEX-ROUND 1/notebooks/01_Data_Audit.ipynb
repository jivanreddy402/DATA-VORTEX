```python
# ============================================
# CHECK FOR NEGATIVE-LIKE CORRUPTION PATTERN
# ============================================

negative = posts[posts["likes"] < 0].copy()

negative["absolute_likes"] = negative["likes"].abs()

print("Negative likes:", len(negative))

print("\nAbsolute value statistics:")
print(negative["absolute_likes"].describe())

print("\nDo absolute values stay within valid range?")
print(
    "All between 0 and 5000:",
    negative["absolute_likes"].between(0, 5000).all()
)

print("\nFirst 20 corrupted values:")
print(
    negative[
        ["post_id", "likes", "absolute_likes", "shares", "comments"]
    ].head(20)
)
```

```python
# ============================================
# RECOVER SIGN-FLIPPED LIKES
# ============================================

# Preserve original values for auditability
posts["likes_original"] = posts["likes"]

# Flag negative likes before correction
posts["likes_was_negative"] = posts["likes"] < 0

# Recover negative likes using absolute value
posts.loc[posts["likes"] < 0, "likes"] = (
    posts.loc[posts["likes"] < 0, "likes"].abs()
)

print("Negative likes recovered:", posts["likes_was_negative"].sum())

print("\nRemaining negative likes:")
print((posts["likes"] < 0).sum())

print("\nLikes statistics after recovery:")
print(posts["likes"].describe())
```

```python
# ============================================
# INVESTIGATE MISSING LIKES PATTERN
# ============================================

likes_missing = posts["likes"].isna()

print("Total rows:", len(posts))
print("Missing likes:", likes_missing.sum())
print("Missing likes %:", round(likes_missing.mean() * 100, 2))

print("\nPlatform-wise likes analysis:")
platform_likes = posts.groupby(
    "platform",
    dropna=False
)["likes"].agg(
    total_rows="size",
    missing_likes="apply"
)

print(platform_likes)
```

```python
# ============================================
# PLATFORM-WISE MISSING LIKES
# ============================================

platform_likes = posts.groupby(
    "platform",
    dropna=False
)["likes"].agg(
    total_rows="size",
    missing_likes=lambda x: x.isna().sum()
)

platform_likes["missing_rate_%"] = (
    platform_likes["missing_likes"]
    / platform_likes["total_rows"] * 100
).round(2)

print(platform_likes)
```

```python
# ============================================
# INVESTIGATE MISSING TEXT CONTENT
# ============================================

text_missing = posts["text_content"].isna()

print("Total rows:", len(posts))
print("Missing text:", text_missing.sum())
print("Missing text %:", round(text_missing.mean() * 100, 2))

print("\nMissing text by platform:")
print(
    posts.groupby("platform", dropna=False)["text_content"]
    .agg(
        total_rows="size",
        missing_text=lambda x: x.isna().sum()
    )
)

print("\nSample rows with missing text:")
print(
    posts[text_missing][
        ["post_id", "user_id", "platform",
         "likes", "shares", "comments", "timestamp"]
    ].head(15)
)
```

```python
# ============================================
# TIMESTAMP FORMAT ANALYSIS
# ============================================

print("Timestamp missing:", posts["timestamp"].isna().sum())
print("Total timestamps:", posts["timestamp"].notna().sum())

# Detect timestamp format
def detect_timestamp_format(value):
    if pd.isna(value):
        return "Missing"
    
    value = str(value).strip()
    
    if value.isdigit():
        return "Unix"
    elif "T" in value:
        return "ISO"
    elif "-" in value:
        return "DMY"
    else:
        return "Other"

timestamp_formats = posts["timestamp"].apply(
    detect_timestamp_format
)

print("\nTimestamp formats:")
print(timestamp_formats.value_counts())
```

```python
# ============================================
# TEST TIMESTAMP PARSING
# ============================================

def parse_timestamp(value):
    if pd.isna(value):
        return pd.NaT
    
    value = str(value).strip()
    
    # Unix timestamp
    if value.isdigit():
        return pd.to_datetime(
            int(value),
            unit="s",
            errors="coerce"
        )
    
    # ISO format
    if "T" in value:
        return pd.to_datetime(
            value,
            errors="coerce"
        )
    
    # DMY format
    return pd.to_datetime(
        value,
        format="%d-%m-%Y",
        errors="coerce"
    )

parsed_timestamps = posts["timestamp"].apply(parse_timestamp)

print("Invalid/unparseable timestamps:")
print(parsed_timestamps.isna().sum())

print("\nDate range:")
print("Minimum:", parsed_timestamps.min())
print("Maximum:", parsed_timestamps.max())
```

```python
print("\nSample converted timestamps:")
print(
    pd.DataFrame({
        "original": posts["timestamp"],
        "parsed": parsed_timestamps
    }).head(20)
)
```

```python
# ============================================
# DUPLICATE POST_ID VALIDATION
# ============================================

duplicate_ids = posts[
    posts["post_id"].duplicated(keep=False)
]["post_id"]

print("Rows with duplicated post_id:", len(duplicate_ids))
print("Unique duplicated post_ids:", duplicate_ids.nunique())

# Check whether duplicated post_ids have identical rows
duplicate_groups = posts[
    posts["post_id"].isin(duplicate_ids)
].groupby("post_id").nunique()

print("\nMaximum number of different values within a duplicated post_id:")
print(duplicate_groups.max())

print("\nAre duplicated post_id groups identical?")
print((duplicate_groups <= 1).all())
```

```python
# ============================================
# REMOVE EXACT DUPLICATE RECORDS
# ============================================

before_duplicates = len(posts)

posts = posts.drop_duplicates().copy()

after_duplicates = len(posts)

print("Rows before duplicate removal:", before_duplicates)
print("Rows after duplicate removal:", after_duplicates)
print("Duplicate rows removed:", before_duplicates - after_duplicates)

print("\nRemaining duplicate rows:")
print(posts.duplicated().sum())

print("\nRemaining duplicated post_ids:")
print(posts["post_id"].duplicated().sum())
```

```python
# ============================================
# STANDARDIZE TIMESTAMPS
# ============================================

posts["timestamp_original"] = posts["timestamp"]

posts["timestamp_clean"] = posts["timestamp"].apply(parse_timestamp)

print("Invalid timestamps after parsing:",
      posts["timestamp_clean"].isna().sum())

print("\nClean timestamp range:")
print("Minimum:", posts["timestamp_clean"].min())
print("Maximum:", posts["timestamp_clean"].max())

print("\nTimestamp sample:")
print(
    posts[["timestamp_original", "timestamp_clean"]].head(10)
)
```

```python
# ============================================
# CLEAN HTML-ENCODED TEXT
# ============================================

from html import unescape

posts["text_content_original"] = posts["text_content"]

posts["text_content"] = posts["text_content"].apply(
    lambda x: unescape(x) if pd.notna(x) else x
)

print("HTML entity occurrences remaining:")
print(posts["text_content"].str.contains(
    "&amp;",
    na=False
).sum())

print("\nSample cleaned text:")
print(
    posts[["text_content_original", "text_content"]]
    .loc[
        posts["text_content_original"].str.contains(
            "&amp;",
            na=False
        )
    ]
    .head(10)
)
```

```python
# ============================================
# CLEANING VALIDATION
# ============================================

print("Final rows:", len(posts))
print("Final columns:", len(posts.columns))

print("\nMissing values:")
print(posts.isna().sum())

print("\nNegative likes remaining:",
      (posts["likes"] < 0).sum())

print("\nDuplicate rows remaining:",
      posts.duplicated().sum())

print("\nDuplicate post_ids remaining:",
      posts["post_id"].duplicated().sum())
```

```python
# ============================================
# FINAL CLEANING VALIDATION
# ============================================

print("=" * 60)
print("FINAL DATASET VALIDATION")
print("=" * 60)

print("\nDataset shape:")
print(posts.shape)

print("\nDuplicate rows:")
print(posts.duplicated().sum())

print("\nDuplicate post_ids:")
print(posts["post_id"].duplicated().sum())

print("\nNegative likes:")
print((posts["likes"] < 0).sum())

print("\nInvalid timestamps:")
print(posts["timestamp_clean"].isna().sum())

print("\nRemaining HTML entities:")
print(
    posts["text_content"]
    .str.contains("&amp;", na=False)
    .sum()
)

print("\nMissing values:")
print(posts.isna().sum())

print("\nFinal columns:")
print(posts.columns.tolist())
```

```python
# ============================================
# EXPORT CLEANED DATASET - AUDIT VERSION
# ============================================

from pathlib import Path

OUTPUT_DIR = BASE_DIR / "data" / "processed"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

AUDIT_OUTPUT = OUTPUT_DIR / "posts_cleaned_audit.csv"

posts.to_csv(AUDIT_OUTPUT, index=False)

print("Saved successfully!")
print("File:", AUDIT_OUTPUT)
print("Rows:", len(posts))
print("Columns:", len(posts.columns))
```
